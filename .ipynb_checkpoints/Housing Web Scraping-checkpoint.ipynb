```python
from bs4 import BeautifulSoup
from selenium import webdriver
import re
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import time 
```

```python
ON_Cities = ['Ontario', 'toronto-on', 'kitchener-on', 'waterloo-on', 'london-on', 'ottawa-on', 'guelph-on', 'hamilton-on']
BC_Cities = ['British Columbia', 'vancouver-bc', 'victoria-bc', 'nanaimo-bc', 'surrey-bc', 'burnaby-bc', 'coquitlam-bc', 'richmond-bc']
AB_Cities = ['Alberta', 'calgary-ab', 'edmonton-ab', 'lethbridge-ab', 'red-deer-ab', 'airdrie-ab', 'medicine-hat-ab', 'st-albert-ab']
SK_Cities = ['Saskatchewan', 'regina-sk', 'saskatoon-sk', 'prince-albert-sk', 'moose-jaw-sk']
MB_Cities = ['Manitoba', 'winnipeg-mb', 'brandon-mb', 'steinbach-mb']
NB_Cities = ['New Brunswick', 'fredericton-nb', 'moncton-nb', 'saint-john-nb', 'dieppe-nb']
```

```python
## Patterns that we are looking for
# pattern for price
pricePat = r'\$[0-9,]+<'
# pattern for bedrooms
bedPat = r'>[0-9 \+]+ bd</li'
# pattern for bathrooms
bathPat = r'>[0-9 \+]+ ba</li'
# pattern for square feet
sqftPat = r'[0-9][0-9 -]+ sf</li'
# pattern for property type
ptPat = r'(?!.*wrap">).*?</div></section>'


## Helpful function for turning price strings into integers
def priceRead(prc):
    prc = prc.replace(',', '')
    return int(prc)

## Helpful function for turning bed/bath strings into integers
def bedRead(bd):
    if '+' in bd:
        # remove spaces
        bd = bd.replace(' ', '')
        # return sum of numbers
        bd = bd.split('+')
        return int(bd[0]) + int(bd[1])
    else:
        return int(bd)

## Helpful function for turning sqft strings into values
def sqftRead(sq):
    if '-' in sq:
        # remove spaces
        sq = sq.replace(' ', '')
        # return max of numbers
        sq = sq.split('-')
        return int(sq[1])
    else:
        return int(sq)

## Initiate houses list
houses = []
```

```python


def CityListScraper(cityList):
    dr = webdriver.Chrome()
    baseURL = 'https://www.rew.ca/properties/areas/'
    prevLen = 0
    houses = []
    
    for city in cityList[1:]:
        print("Scraping", city, "...")
        pageNum = 1
        while (True):
            # get new URL
            url = baseURL + city + '/page/' + str(pageNum)
            
            # and scrape it to turn it into a helpful list
            dr.get(url)
            soup = BeautifulSoup(dr.page_source, "html.parser")
            houseList = soup.find_all('div', {'class':'displaycard-body'})
            
            # check if we have gone past last page
            if soup.find('div', {'class':'pageblock-cta_wrapper'}):
                break
        
            # go through each house in the list
            for index, house in enumerate(houseList):
                house = str(house)
                price = re.search(pricePat, house)
                if price:
                    price = priceRead(price.group()[1:-1])
                bed = re.search(bedPat, house)
                if bed:
                    bed = bedRead(bed.group()[1:-7])
                bath = re.search(bathPat, house)
                if bath:
                    bath = bedRead(bath.group()[1:-7])
                sqft = re.search(sqftPat, house)
                if sqft:
                    sqft = sqftRead(sqft.group()[0:-7])
                prop = re.search(ptPat, house)
                if prop:
                    prop = prop.group()[5:-16]
                data = {
                    "Province": cityList[0],
                    "City": city,
                    "Price": price,
                    "Bedrooms": bed,
                    "Bathrooms": bath,
                    "Square Feet": sqft,
                    "Property Type": prop
                }
                if sqft == 0:
                    print(house,'\n')
                houses.append(data)
            
            # only proceed to next page if current page was properly scraped
            if len(houses) > prevLen:
                print("      Scraped page", pageNum, "| total listings:", len(houses))
                #print(houses)
                pageNum += 1
                prevLen = len(houses)
        

    # After everything, return the list
    return houses
```

```python
ON_df = pd.DataFrame(CityListScraper(ON_Cities))
```

```python
BC_df = pd.DataFrame(CityListScraper(BC_Cities))
```

```python
AB_df = pd.DataFrame(CityListScraper(AB_Cities))
```

```python
SK_df = pd.DataFrame(CityListScraper(SK_Cities))
```

```python
MB_df = pd.DataFrame(CityListScraper(MB_Cities))
```

```python
NB_df = pd.DataFrame(CityListScraper(NB_Cities))
```

```python
housing_df = pd.concat([ON_df, BC_df, AB_df, SK_df, MB_df, NB_df], ignore_index=True)
housing_df
```

```python
## let's save the dataframe as a csv file
housing_df.to_csv("housing_df.csv", index=False)
```
